#     Implement and investigate the exponential growth equation.  

In the year 600 the human population had a size of 200 million people. In the year 1300 the population size was 360 million.   

-Calculate the per capita growth rate per year. Do this under assumption that the population grows exponentially.  

-Calculate the doubling time of the human population based on these data, assuming that the population grows exponentially.  

```python
# per capita
import numpy as np


T1 = 600
T2 = 1300
N_T2 = 360
N_T1 = 200

r = (1 / (T2 - T1)) * np.log(N_T2 / N_T1)
r

# Doubling time

t_double = np.log(2) / r
t_double
```

The human population had 265 million people in the year 1000. In the year 1200 it had reached a size of 360 million. in this question we will forecast the growth based on these two data points.  

-If you assume that the human population grew exponentially and would have continued to grow exponentially with the same rate, how big would the population be in 1975? You can use your maths skills to answer the question. It might be easier to implement the exponential growth function in Python and find the answer with your code.  

```python
import numpy as np


T1 = 1000
T2 = 1200
N_T2 = 360
N_T1 = 265

r = (1 / (T2 - T1)) * np.log(N_T2 / N_T1)
r

# assuming exponential growth, predict population at T=1975
N_pred = N_T2 * np.exp(r * (1975 - T2))
N_pred
```
